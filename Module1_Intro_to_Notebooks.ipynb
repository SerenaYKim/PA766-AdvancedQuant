## 1. What is a notebook?

A Jupyter notebook is made of **cells**. There are two kinds:

- **Markdown cells** (like this one) hold text, notes, headers, and explanations. They don't run — they just display formatted text.
- **Code cells** hold actual Python code. When you run one, it executes and shows its output directly underneath.

**To run a cell:** click on it, then press **Shift + Enter** (runs the cell and moves to the next one), or click the ▶ Run button.

**Important:** cells run in whatever order *you* run them, not necessarily top to bottom. The number in brackets next to a code cell (like `[1]`) tells you the order it was executed. If something errors and you're confused, a good fix is: **Kernel → Restart Kernel and Run All Cells** — this reruns everything from scratch, top to bottom, cleanly.

## 3. Loading our first dataset

Today's dataset is real: **state-level voter turnout in U.S. presidential elections, 1920–2012**, along with whether each state had three specific election-administration policies in place that year — Election Day (same-day) registration, mail-in/absentee voting, and "motor voter" registration.

Make sure `turnout.csv` is saved in the **same folder** as this notebook, then run:

```python
df = pd.read_csv("turnout.csv")
df.head()
```

`.head()` shows the first 5 rows — a quick sanity check that the data loaded correctly and looks like what you expect.

## 4. Getting oriented

Before doing anything else with a new dataset, it's worth answering three questions: *How big is it? What type is each column? What does the data look like overall?*

```python
df.shape   # (number of rows, number of columns)
```

```python
df.info()   # column names, data types, missing values
```

```python
df.describe(include="all")   # quick summary statistics for every column
```

| Column | Likely meaning |
|---|---|
| `abb` | Two-letter U.S. state postal abbreviation (e.g., `AL`, `MN`). 47 states appear in this file. |
| `year` | Presidential election year. Every 4 years, 1920–2012 (24 elections total). |
| `turnout` | State-level voter turnout rate — the percentage of the voting-age population that voted in that state's presidential election that year. |
| `policy_edr` | 1 if the state had **Election Day / same-day voter registration** in effect that year, 0 otherwise. |
| `policy_mail_in` | 1 if the state allowed some form of **mail-in / absentee voting** that year, 0 otherwise. |
| `policy_motor` | 1 if the state had a **"Motor Voter"** law (voter registration offered at the DMV) in effect that year, 0 otherwise. |

## 5. Looking at one state over time

pandas lets you filter rows with a condition inside square brackets. Let's pull out just Minnesota and sort it by year.

```python
minnesota = df[df["abb"] == "MN"].sort_values("year")
minnesota
```

A table of numbers is hard to read at a glance — let's plot it instead.

```python
plt.figure(figsize=(9, 5))
plt.plot(minnesota["year"], minnesota["turnout"], marker="o")
plt.xlabel("Year")
plt.ylabel("Turnout (%)")
plt.title("Voter turnout in Minnesota, 1920–2012")
plt.show()
```

## 6. Comparing groups — a first taste of this semester's central question

Here's a question a policymaker might actually ask: **does Election Day registration raise turnout?**

A first, very rough pass: compare the *average* turnout in state-years that had the policy vs. state-years that didn't.

```python
df.groupby("policy_edr")["turnout"].mean()
```

You'll likely see a noticeably higher average where `policy_edr = 1`.

**Does that mean EDR *causes* higher turnout?** Not necessarily — and figuring out why not is a big part of what this course is about.

States that chose to adopt EDR (several upper-Midwest and New England states, historically) may have already been states with unusually civic-minded populations, strong local political cultures, or other factors that would have produced high turnout *regardless* of the policy. A simple group average can't tell the difference between "the policy worked" and "the kind of state that adopts this policy was already going to have high turnout."

This is the difference between **description** (what does the data show?), **prediction** (can I guess turnout from these variables?), and **inference** (does X actually *cause* Y?) — three different questions that need different tools. We'll spend the semester building those tools properly, starting next class.

A bar chart makes the same comparison a little easier to read at a glance:

```python
group_means = df.groupby("policy_edr")["turnout"].mean()
group_means.plot(kind="bar")
plt.xlabel("Election Day Registration in effect (0 = No, 1 = Yes)")
plt.ylabel("Average turnout (%)")
plt.title("Average turnout by EDR policy status")
plt.xticks(rotation=0)
plt.show()
```

## 7. Looking at the whole distribution

So far we've looked at averages. But averages hide a lot — two states could have the same average turnout with very different amounts of variation around it. A histogram shows the full spread.

```python
plt.figure(figsize=(8, 5))
df["turnout"].hist(bins=30, edgecolor="white")
plt.xlabel("Turnout (%)")
plt.ylabel("Number of state-year observations")
plt.title("Distribution of turnout across all state-years, 1920–2012")
plt.show()
```
